```python
import pandas as pd
import pickle
from sklearn.linear_model import LogisticRegression
```

```python
df=pd.read_csv(r"C:\Users\Manimozhi\Downloads\iris_data.csv")
df.shape
```

```python
x=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
model=LogisticRegression().fit(x,y)
```

```python
x.loc[0]
```

```python
model.predict([0.1,0.2,1.0,3.2])
```

```python
#saving model to local device

with open('iris-model.pkl','wb') as f:
    pickle.dump(model,f)
```
